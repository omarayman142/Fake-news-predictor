```python
#import pandas library
import pandas as pd
import numpy as np
from sklearn.metrics import confusion_matrix , classification_report
from sklearn.model_selection import train_test_split

#mix true and fake csv's datasets and make true article with label 1 and fake with 0
df_true = pd.read_csv('True.csv')
df_true['label'] = 1
df_fake = pd.read_csv('Fake.csv')
df_fake['label'] = 0
df = pd.concat([df_true, df_fake])
df.reset_index(drop=True, inplace=True)
df.head()



```

```python
#make them random
df = df.sample(frac=1)
df.reset_index(drop=True, inplace=True)
df.head()
```

```python
df
```

```python
df=df.drop(["subject","date"],axis=1)
df.head()
```

```python
import spacy

# Install spaCy if not already installed and download the English model
# You might need to restart the runtime after installation if it's the first time
# !pip install spacy

try:
    nlp = spacy.load('en_core_web_lg')
except OSError:
    print('Downloading spaCy large model. This may take a moment...')
    !python -m spacy download en_core_web_lg
    nlp = spacy.load('en_core_web_lg')
```

```python
# Function to remove stop words
def remove_stopwords(text):
    if pd.isna(text):
        return text
    doc = nlp(text)
    tokens = [token.text for token in doc if not token.is_stop]
    return ' '.join(tokens)

# Apply the function to the 'title' and 'text' columns
df['title_processed'] = df['title'].apply(remove_stopwords)
df['text_processed'] = df['text'].apply(remove_stopwords)

print("Stop words removed from 'title' and 'text' columns. New columns 'title_processed' and 'text_processed' created.")
df.head()
```

```python
df['combined_text'] = df['title_processed'] + ' ' + df['text_processed']
df.head()
```

```python
# Function to perform lemmatization
def lemmatize_text(text):
    if pd.isna(text):
        return text
    doc = nlp(text)
    tokens = [token.lemma_ for token in doc]
    return ' '.join(tokens)

# Apply the function to the 'combined_text' column
df['combined_text_lemmatized'] = df['combined_text'].apply(lemmatize_text)

print("Lemmatization applied to 'combined_text' column. New column 'combined_text_lemmatized' created.")
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer with min_df and max_df
# min_df=5: Ignores terms that appear in less than 5 documents.
# max_df=0.8: Ignores terms that appear in more than 80% of the documents.
vectorizer = CountVectorizer(min_df=5, max_df=0.8,ngram_range=(2,3))

# Fit and transform the 'combined_text' column
bow_matrix = vectorizer.fit_transform(df['combined_text'])

# Display the shape of the Bag-of-Words matrix
print("Shape of Bag-of-Words matrix:", bow_matrix.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TfidfVectorizer with similar parameters to CountVectorizer for consistency
# You can adjust min_df, max_df, and ngram_range as needed
tfidf_vectorizer = TfidfVectorizer(min_df=5, max_df=0.8, ngram_range=(1,4))

# Fit and transform the 'combined_text_lemmatized' column
tfidf_matrix = tfidf_vectorizer.fit_transform(df['combined_text_lemmatized'])

# Display the shape of the TF-IDF matrix
print("Shape of TF-IDF matrix:", tfidf_matrix.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(tfidf_matrix, df['label'], test_size=0.2, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from sklearn.naive_bayes import MultinomialNB

# Initialize the Multinomial Naive Bayes classifier
nb_classifier = MultinomialNB()

# Train the classifier using the training data
nb_classifier.fit(X_train, y_train)

print("Multinomial Naive Bayes classifier trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Make predictions on the test set
y_pred = nb_classifier.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

# Generate classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Generate confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
